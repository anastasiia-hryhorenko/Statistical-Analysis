Завдання 1. Підрахунок загальної кількості респондентів

```python
respondents_count = df['ResponseId'].nunique()
print(f'Загальна кількість респондентів: {respondents_count}')
```

Завдання 2. Аналіз повноти відповідей респондентів

```python
#Сиворюємо множину зі списку питань з таблиці schema
df1_questions = set(df1['qname'])
```

```python
#Сиворюємо множину зі стовбців з таблиці results
df_columns = set(df.columns)
```

```python
#робимо інтерсекцію цих двох множин(знаходимо спільні питання)
questions_intersection = list(df1_questions.intersection(df_columns))
```

```python
#Очищаємо тримані дані від пропущених значень, тобто лишають тільки ті респонденти у яких не має пропущених значень у відповідях
questions_cleaned = df[questions_intersection].dropna()
```

```python
#Підраховуємо кількість таких респондентів, виводимо відповідь
print('Кількість респондентів, що відповіли на всі запитання опитування:', len(questions_cleaned))
```

Завдання 3. Статистичний аналіз досвіду респондентів

```python
#Перевіряємо тип даних в колонці WorkExp
print(df['WorkExp'].dtype)
```

```python
print('Міри центральної тенденції:')

#Рахуємо середнє значення, округлюємо до одного знаку після коми
avg = round(df['WorkExp'].mean(), 1)
#Рахуємо медіану
median = df['WorkExp'].median()
#Рахуємо моду
mode = df['WorkExp'].mode()

#Виводимо результат
print(f'Середнє значення: {avg}')
print(f'Медіана: {median}')
print(f'Мода: {mode}')
```

Завдання 5. Визначення популярності Python

```python
#створюємо стовпець з булевими значеннями для тих респондентів, хто програмує на Python
df['worked_with_python'] = df['LanguageHaveWorkedWith'].str.contains('python', case=False, na=False)

print(df['worked_with_python'])
```

```python
#Визначаємо відсоток таких респондентів
respondents_perc = round(df['worked_with_python'].mean()*100, 1)
```

```python
print(f'Відсоток респондентів: {respondents_perc}%')
```

Завдання 6. Аналіз шляхів навчання програмуванню

```python
#створюємо стовпець з булевими значеннями для тих респондентів, хто навчався на онлайн курсах
df['learned_with_online_courses'] = df['LearnCode'].str.contains('online courses', case=False, na=False)
df['learned_with_online_courses']
```

Завдання 7. Географічний аналіз компенсації Python-розробників

```python
#Обираємо респодентів, які програмують на Python
df_new = df[df['worked_with_python']]
```

```python
#видаляємо пропущені значення з колонки ConvertedCompYearly
df_new = df_new.dropna(subset=['ConvertedCompYearly'])
```

```python
#Робимо групування по країнам, рахуємо середнє та медіанту, округлюємо числа до 2 знаків після коми, індекси перетвоюємо на колонку
geo_analysis = df_new.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median']).round(2).reset_index()
#Перейменовуємо колонки
geo_analysis.columns = ['Country', 'Avg Comp', 'Med Comp']
#виводимо результат
geo_analysis
```

Завдання 8. Аналіз освіти найбільш оплачуваних спеціалістів

```python
#Для зручнішої візуалізації формуємо нову таблицю з колонок, які будемо використовувати, 
#сортуємо колонку ConvertedCompYearly поспаданню, впорядковуємо індекси, виводимо перші 5 респондентів
df_education = df[['ResponseId', 'EdLevel', 'ConvertedCompYearly']].sort_values('ConvertedCompYearly', ascending=False).reset_index(drop=True).head(5)
df_education
```

Завдання 9. Аналіз популярності Python по віковим категоріям 

```python
#визначаємо % респондентів, що програмують на Python, групуючи по віковим категоріям
age_group = df.groupby('Age')['worked_with_python'].mean()*100
#Перетворюємо індекси на звичайну колонку
age_group = age_group.reset_index()
#Округлюємо відсотки до одного знаку після коми
age_group['worked_with_python'] = age_group['worked_with_python'].round(1)
#Перейменовуємо колонки
age_group.columns = ['Age Group', 'Python Users, %']
#видодимо таблицю з результатом
age_group
```

Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників

```python
#Для зручнішої візуалізації формуємо нову таблицю з колонок, які будемо використовувати. 
df_10 = df[['ResponseId', 'ConvertedCompYearly', 'RemoteWork', 'Industry']].sort_values('ConvertedCompYearly', ascending=False).reset_index(drop=True)
```

```python
#Знаходимо 75 персентиль за компенсацією
perc = df_10['ConvertedCompYearly'].quantile(0.75)
print(perc)
```

```python
#Фільтруємо респондкнтів по двум умовам: їхня оплата більша від 75% респондентів та працюють віддалено
filtered = df_10[(df_10['ConvertedCompYearly'] > perc) & (df_10['RemoteWork'] == 'Remote')]
```

```python
#Виводимо список найпопулярніших індустрій серед високооплачуваних працівників
popular_inductries = df_10['Industry'].value_counts()
print('Найпопулярніші індустрії серед високооплачуваних працівників:')
print(popular_inductries)
```
